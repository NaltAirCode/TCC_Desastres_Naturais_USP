```python
# ==============================================================================
# TESTE DE CONEXÃO E INSPEÇÃO DO FORMULÁRIO DE RELATÓRIOS DO S2iD (MIDR)
# URL Oficial: https://s2id.mi.gov.br/paginas/relatorios/
# ==============================================================================
import requests
from bs4 import BeautifulSoup
import urllib3

# Desabilita avisos de certificado SSL governamental (comum em portais .gov.br)
urllib3.disable_warnings(urllib3.exceptions.InsecureRequestWarning)

URL_S2ID = "https://s2id.mi.gov.br/paginas/relatorios/"

sessao = requests.Session()
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/124.0.0.0 Safari/537.36"
}

print(f"Conectando ao portal S2iD: {URL_S2ID} ...")
try:
    resposta = sessao.get(URL_S2ID, headers=headers, verify=False, timeout=30)
    print(f"Status HTTP: {resposta.status_code}")
    
    soup = BeautifulSoup(resposta.text, "html.parser")
    
    # Verifica se há formulários JSF/ViewState ou iframes de painel (Qlik/BI)
    viewstate = soup.find("input", {"name": "javax.faces.ViewState"})
    iframes = soup.find_all("iframe")
    botoes = soup.find_all(["button", "a", "input"])
    
    print(f"Token ViewState encontrado? {'Sim' if viewstate else 'Não'}")
    print(f"Quantidade de iframes na página: {len(iframes)}")
    for idx, iframe in enumerate(iframes):
        print(f"  -> Iframe {idx+1} src: {iframe.get('src')}")
        
    print("\nPrincipais elementos/links de relatórios identificados no HTML:")
    for b in botoes[:15]:
        texto = b.get_text(strip=True) or b.get("value") or b.get("id")
        if texto:
            print(f"  -> Tag <{b.name}> | ID: {b.get('id')} | Texto: {texto[:60]}")
            
except Exception as e:
    print(f"Erro ao conectar no S2iD a partir do cluster Databricks: {e}")
```
