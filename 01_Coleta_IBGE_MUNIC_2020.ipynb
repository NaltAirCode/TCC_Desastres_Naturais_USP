```python
import os
import requests
from tqdm import tqdm

# 1. Cole aqui o caminho do Volume que você copiou no Databricks
CAMINHO_VOLUME = "/Volumes/workspace/default/fontes_tcc"

# 2. O link exato que você encontrou no servidor do IBGE
url_munic = "https://ftp.ibge.gov.br/Perfil_Municipios/2020/Base_de_Dados/Base_MUNIC_2020.xlsx"
arquivo_destino = os.path.join(CAMINHO_VOLUME, "Base_MUNIC_2020.xlsx")

print(f"Tentando acessar: {url_munic}")
resposta = requests.get(url_munic, stream=True, timeout=120)

if resposta.status_code == 200:
    print(f"URL encontrada: {url_munic}")
    tamanho_total = int(resposta.headers.get("content-length", 0))
    
    with open(arquivo_destino, "wb") as f, tqdm(
        desc="Baixando: Base_MUNIC_2020.xlsx",
        total=tamanho_total,
        unit="B",
        unit_scale=True,
        unit_divisor=1024
    ) as barra:
        for bloco in resposta.iter_content(chunk_size=8192):
            f.write(bloco)
            barra.update(len(bloco))
            
    print(f"\n[SUCESSO] Arquivo salvo no Volume: {arquivo_destino}")
else:
    print(f"[ERRO] Código HTTP: {resposta.status_code}")
```
